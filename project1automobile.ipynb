```python
df_auto.replace('?',np.nan,inplace = True) 
df_auto
```

```python
# change the data tyoes 

df_auto['price'] =df_auto['price'].apply(float)                     # also use stype
df_auto['normalized-losses'] =df_auto['normalized-losses'].apply(float) 
df_auto['horsepower'] =df_auto['horsepower'].apply(float)    
df_auto['peak-rpm'] =df_auto['peak-rpm'].apply(float)
```

```python
df_auto.info()
```

```python
df_auto.duplicated().sum()
```

```python
df_auto.isnull().sum()
```

```python
avg_normalized_losses = df_auto['normalized-losses'].mean()
df_auto['normalized-losses'].replace(np.nan,avg_normalized_losses,inplace = True) 

avg_price = df_auto['price'].mean()
df_auto['price'].replace(np.nan,avg_price,inplace = True)
```

```python
df_auto.isnull().sum()
```

```python
fre_door = df_auto['num-of-doors'].dropna().mode()        
fre_door                          # ans is in series struture we need to acces that value so i give o index 
```

```python
fre_door = df_auto['num-of-doors'].dropna().mode()[0]      
fre_door
```

```python
fre_door = df_auto['num-of-doors'].dropna().mode()[0]  
df_auto['num-of-doors'].replace(np.nan,fre_door,inplace=True)


fre_bore = df_auto['bore'].dropna().mode()[0]  
df_auto['bore'].replace(np.nan,fre_bore,inplace=True)

fre_stroke = df_auto['stroke'].dropna().mode()[0]  
df_auto['stroke'].replace(np.nan,fre_stroke,inplace=True)

fre_horsepower= df_auto['horsepower'].dropna().mode()[0]  
df_auto['horsepower'].replace(np.nan,fre_horsepower,inplace=True)


fre_peak_rpm= df_auto['peak-rpm'].dropna().mode()[0]  
df_auto['peak-rpm'].replace(np.nan,fre_peak_rpm,inplace=True)
```
