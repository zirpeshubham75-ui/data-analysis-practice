# Business Problem: Employee Attrition Analysis 

# Q1. What is the overall employee attrition rate?
👨‍💼 Interviewer:

# "Can you tell me what percentage of employees have left the company?"

🧑‍💻 Your answer:

"Yes. I would calculate the mean of the LeaveOrNot column because it contains binary values, where 1 represents employees who left and 0 represents employees who stayed. Multiplying the mean by 100 gives me the attrition percentage."

```python
 df['LeaveOrNot'].mean() * 100 
```

# Q2. Which city has the highest employee attrition rate?
👨‍💼 Interviewer:

"Which city has the highest employee attrition rate?"

🧑‍💻 Your answer:

"I would group the employees by city and calculate the mean of LeaveOrNot. Since LeaveOrNot is binary, the mean represents the attrition rate."

```python
df.groupby(by='City')['LeaveOrNot'].mean()
```

# Q3. Which Payment Tier has the highest attrition?
👨‍💼 Interviewer:

"Can you identify which payment tier has the highest employee attrition?"

🧑‍💻 Your answer:

"Yes. I would group employees by payment tier and calculate the attrition rate for each tier."

```python
df.groupby(by='PaymentTier')['LeaveOrNot'].mean()
```

```python
df.groupby(by='City')['LeaveOrNot'].mean().sort_values(ascending=False)
```
